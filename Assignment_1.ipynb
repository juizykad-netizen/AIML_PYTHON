```python
#ATM CALCULATOR
balance = float(input("Enter account balance: "))
withdraw = float(input("Enter withdrawal amount: "))

if withdraw <= balance:
    balance -= withdraw
    print(f"Transaction successful. Remaining balance: ₹{balance}")
else:
    print("Error: Insufficient balance.")
```

```python
#ELECTRICITY BILL GENERATOR
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + (units - 100) * 7
else:
    bill = (100 * 5) + (100 * 7) + (units - 200) * 10

print(f"Bill amount = ₹{bill}")
```

```python
#TEMPERATURE CONVERTER
temp = float(input("Enter temperature value: "))
unit = input("Enter unit (C/F): ").upper()

if unit == "C":
    f = (temp * 9/5) + 32
    print(f"Temperature in Fahrenheit: {f}°F")
elif unit == "F":
    c = (temp - 32) * 5/9
    print(f"Temperature in Celsius: {c}°C")
else:
    print("Invalid unit.")
```

```python
#MOBILE RECHARGE VALIDITY
from datetime import datetime, timedelta

recharge_date = input("Enter recharge date (YYYY-MM-DD): ")
today_date = input("Enter today's date (YYYY-MM-DD): ")

r = datetime.strptime(recharge_date, "%Y-%m-%d")
t = datetime.strptime(today_date, "%Y-%m-%d")

expiry = r + timedelta(days=28)

if t <= expiry:
    print("Plan is still valid.")
else:
    print("Plan expired. Recharge again.")
```

```python
#6	  SUM OF EVEN NUMBERS
n = int(input("Enter a number: "))
total = 0

for i in range(2, n+1, 2):
    total += i

print(f"Sum of even numbers: {total}")
```

```python
#7	  STUDENT MARKS SUMMARY
students = {}

for i in range(3):
    name = input("Enter name: ")
    marks = float(input("Enter marks: "))
    students[name] = marks

avg = sum(students.values()) / 3
print(f"Average Marks: {avg}")
```

```python
#8	  WORD FREQUENCY COUNTER
sentence = input("Enter a sentence: ")
words = sentence.split()

freq = {}

for w in words:
    freq[w] = freq.get(w, 0) + 1

for word, count in freq.items():
    print(f"{word}: {count}")
```

```python
#9	  PRIME NUMBER CHECKER
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

n = int(input("Enter a number: "))

if is_prime(n):
    print(f"{n} is a prime number.")
else:
    print(f"{n} is not a prime number.")
```

```python
#10. write a python program that takes a string and returns no of vovels
# Input from user
text = input("Enter a string: ")

# Initialize vowel count
vowel_count = 0

# List of vowels
vowels = "aeiouAEIOU"

# Loop through each character in the string
for char in text:
    if char in vowels:
        vowel_count += 1

# Display the result
print("Number of vowels in the string:", vowel_count)
```

```python
#11. Write a program that takes two numbers as input: a range limit N and a divisor D. Count how many numbers between 1 and N are divisible by D.
# Take inputs
N = int(input("Enter the range limit N: "))
D = int(input("Enter the divisor D: "))

# Initialize count
count = 0

# Loop through numbers from 1 to N
for i in range(1, N+1):
    if i % D == 0:
        count += 1

# Display result
print("Number of integers between 1 and", N, "divisible by", D, "is:", count)
```
