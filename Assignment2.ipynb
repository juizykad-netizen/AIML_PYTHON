```python
#ATM CALCULATOR
balance = float(input("Enter account balance: "))
withdraw = float(input("Enter withdrawal amount: "))

if withdraw <= balance:
    balance -= withdraw
    print(f"Transaction successful. Remaining balance: ₹{balance}")
else:
    print("Error: Insufficient balance.")
```

```python
#ELECTRICITY BILL GENERATOR
units = int(input("Enter units consumed: "))

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = (100 * 5) + (units - 100) * 7
else:
    bill = (100 * 5) + (100 * 7) + (units - 200) * 10

print(f"Bill amount = ₹{bill}")
```

```python
#TEMPERATURE CONVERTER
temp = float(input("Enter temperature value: "))
unit = input("Enter unit (C/F): ").upper()

if unit == "C":
    f = (temp * 9/5) + 32
    print(f"Temperature in Fahrenheit: {f}°F")
elif unit == "F":
    c = (temp - 32) * 5/9
    print(f"Temperature in Celsius: {c}°C")
else:
    print("Invalid unit.")
```

```python
#MOBILE RECHARGE VALIDITY
from datetime import datetime, timedelta

recharge_date = input("Enter recharge date (YYYY-MM-DD): ")
today_date = input("Enter today's date (YYYY-MM-DD): ")

r = datetime.strptime(recharge_date, "%Y-%m-%d")
t = datetime.strptime(today_date, "%Y-%m-%d")

expiry = r + timedelta(days=28)

if t <= expiry:
    print("Plan is still valid.")
else:
    print("Plan expired. Recharge again.")
```

```python
#6	  SUM OF EVEN NUMBERS
n = int(input("Enter a number: "))
total = 0

for i in range(2, n+1, 2):
    total += i

print(f"Sum of even numbers: {total}")
```

```python
#7	  STUDENT MARKS SUMMARY
students = {}

for i in range(3):
    name = input("Enter name: ")
    marks = float(input("Enter marks: "))
    students[name] = marks

avg = sum(students.values()) / 3
print(f"Average Marks: {avg}")
```

```python
#8	  WORD FREQUENCY COUNTER
sentence = input("Enter a sentence: ")
words = sentence.split()

freq = {}

for w in words:
    freq[w] = freq.get(w, 0) + 1

for word, count in freq.items():
    print(f"{word}: {count}")
```

```python
#9	  PRIME NUMBER CHECKER
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

n = int(input("Enter a number: "))

if is_prime(n):
    print(f"{n} is a prime number.")
else:
    print(f"{n} is not a prime number.")
```

```python
#3. Load the Iris dataset
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(
    iris.data,
    columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

df.head()
```

```python
#Display first 10 rows and basic statistics
df.head(10)
```

```python
df.describe()
```

```python
#b) Filter rows where sepal_length > median and species is "versicolor"
median_sepal_length = df['sepal_length'].median()

filtered_df = df[
    (df['sepal_length'] > median_sepal_length) &
    (df['species'] == 'versicolor')
]

filtered_df
```

```python
#c) Group by species and compute mean petal_length
mean_petal_length = df.groupby('species')['petal_length'].mean()
mean_petal_length
```

```python
#Add new column petal_ratio = petal_length / petal_width
df['petal_ratio'] = df['petal_length'] / df['petal_width']
df.head()
```

```python
#a) Top 5 rows with highest petal_ratio
df.sort_values(by='petal_ratio', ascending=False).head(5)
```

```python
#Data Visualization (Matplotlib)
#5a. Histogram of sepal_length (15 bins)
plt.figure()
plt.hist(df['sepal_length'], bins=15)
plt.xlabel('Sepal Length')
plt.ylabel('Frequency')
plt.title('Histogram of Sepal Length')
plt.show()
```

```python
#5b. Scatter plot: sepal_length vs petal_length (color-coded by species)
plt.figure()

for species in df['species'].unique():
    subset = df[df['species'] == species]
    plt.scatter(subset['sepal_length'], subset['petal_length'], label=species)

plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.title('Sepal Length vs Petal Length')
plt.legend()
plt.show()
```

```python
#Data Visualization (Seaborn)
#6a. Pairplot of the entire dataset
sns.pairplot(df, hue='species')
plt.show()
```

```python
#6b. Boxplot of petal_width for each species
plt.figure()
sns.boxplot(x='species', y='petal_width', data=df)
plt.title('Petal Width Distribution by Species')
plt.show()
```
